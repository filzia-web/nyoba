Latihan 1


```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

NumPy Array dapat dikalikan dengan angka desimal seperti 0.9 sehingga setiap elemennya otomatis dikalikan. Sedangkan List Python biasa tidak bisa dikalikan dengan angka desimal dan akan menghasilkan error. Hal ini karena NumPy mendukung operasi perhitungan langsung pada setiap elemen array, sedangkan List Python tidak memiliki kemampuan tersebut secara langsung.

Latihan  2

```python
import pandas as pd

data_kantin = {
    "menu": ["Nasi Goreng", "Es Teh", "Mie Ayam", "Es Teh", "None"],
    "harga": [12000, 4000, 10000, 4000, 8000],
    "terjual": [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Kolom yang memiliki data kosong adalah kolom menu dan terjual. Data kosong dapat menyebabkan hasil analisis menjadi tidak akurat karena beberapa nilai tidak dapat dihitung atau tidak diketahui keterangannya. Jadi, missing value perlu ditangani terlebih dahulu sebelum data dianalisis.

Latihan 3


```python
# Menggunakan DataFrame df yang sudah didefinisikan di sel sebelumnya (Latihan 2)
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Kolom terjual memiliki jumlah non null lebih sedikit daripada jumlah baris total. Jumlah non null pada kolom terjual adalah 4, sedangkan jumlah baris total adalah 5. Jadi terdapat 1 data kosong pada kolom terjual.

Latihan 4

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom

df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Kolom terjual diisi dengan 0 menggunakan fillna() karena data kosong dapat dianggap sebagai tidak ada penjualan yang tercatat. Sedangkan baris dengan menu  kosong dihapus dengan dropna() karena nama menu itu informasi penting untuk mengetahui data tersebut termasuk menu apa. Jika dibiarkan, data akan sulit dianalisis berdasarkan menu.

Latihan 5

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Setelah drop_duplicates(), jumlah baris tetap 4 karena tidak terdapat data yang duplikat. Tipe data perlu dipastikan benar agar perhitungan dan analisis data dapat dilakukan dengan tepat serta menghindari kesalahan dalam pengolahan data.

Latihan 6

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting

df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan

ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Menu yang menghasilkan total penapatan tertinggi adalah es teh, dengan jumlah pendapatan sebesar Rp.300.000. Informasi ini dapat membantu kantin sekolah dalam menentukan menu yang paling menguntungkan, sehingga kantin bisa menyediakan stok lebih banyak dan memprioritaskan menu tersebut.
